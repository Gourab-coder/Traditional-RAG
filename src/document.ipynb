```python
# create a simple txt file
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_texts={
    "../data/text_files/python_intro.txt":"""
    Python Programming

Python is a popular high-level programming language used to develop different types of software and applications. It was created by Guido van Rossum and was first released in 1991. Python is known for its simple syntax, which makes it easy for beginners to learn and understand.

Python can be used for many purposes, such as web development, data analysis, artificial intelligence, machine learning, automation, and software development. It provides many built-in libraries and frameworks that make programming easier and faster.

One of the main advantages of Python is its readability. A simple Python program can be written in only a few lines. For example, the following program prints a message:

print("Hello, World!")

Python also supports important programming concepts such as variables, loops, functions, lists, dictionaries, classes, and objects. It is an interpreted language, so programs can be executed without compiling the entire program first.

Python has a large community of developers and many useful libraries, such as NumPy, Pandas, Flask, Django, TensorFlow, and PyTorch. Because of these features, Python is widely used by students, developers, data scientists, and AI engineers.

In conclusion, Python is a simple, powerful, and versatile programming language. Its easy-to-understand syntax and wide range of applications make it an excellent language for beginners as well as professional programmers.
    """
}

for filepath, content in sample_texts.items():
    with open(filepath, 'w', encoding="utf-8") as f:
        f.write(content)

print("sample text files created!")
```

```python
# TextLoader
# from langchain_core.document_loaders import TextLoader ## older version

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document = loader.load()

print(document)
```

```python
# Directory Loader
from langchain_community.document_loaders import DirectoryLoader

# load all the text files from the directory
dir_loader = DirectoryLoader(
    "../data/text_files",
    glob = "**/*.txt", ## pattern to match files
    loader_cls = TextLoader, ## loader class to use
    # loader_kwargs = {'encoding': 'utf-8'},
    # show_progress = False
)

documents = dir_loader.load()
documents
```

to know more diff type of document loader go to langchain document loader

```python
# pdf
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

# load all the pdf files from the directory
dir_loader = DirectoryLoader(
    "../data/pdf",
    glob = "**/*.pdf", ## pattern to match files
    loader_cls = PyMuPDFLoader, ## loader class to use
    # loader_kwargs = {'encoding': 'utf-8'},
    # show_progress = False
)

pdf_documents = dir_loader.load()
pdf_documents
```
